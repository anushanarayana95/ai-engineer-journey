```python
import pandas as pd
```

```python
df = pd.read_csv(
    "/workspaces/ai-engineer-journey/data/survey_results_public.csv",
    engine="python",
    on_bad_lines="skip", index_col="ResponseId"
)
schema_df = pd.read_csv("/workspaces/ai-engineer-journey/data/survey_results_schema.csv", index_col="qname")
```

```python
print(df.isna().sum().sort_values(ascending=False).head(20))  # top 20 columns with missing data
```

```python
# Total rows
print(f"Total rows: {len(df)}")

# % missing per column
missing_pct = (df.isna().sum() / len(df) * 100).sort_values(ascending=False)
print(missing_pct.head(20))
```

```python
threshold = 50
cols_to_drop = missing_pct[missing_pct > threshold].index
df_clean = df.drop(columns=cols_to_drop)

print(f"Columns before: {df.shape[1]}")
print(f"Columns after:  {df_clean.shape[1]}")
```

```python
# Fill text columns with "Unknown"
for col in df_clean.select_dtypes(include="object").columns:
    df_clean[col] = df_clean[col].fillna("Unknown")

# Fill numeric columns with 0
for col in df_clean.select_dtypes(include="number").columns:
    df_clean[col] = df_clean[col].fillna(0)

print(df_clean.isna().sum().sum())  # Should be 0
```

```python
# Fill text columns with "Unknown"
for col in df_clean.select_dtypes(include="str").columns:
    df_clean[col] = df_clean[col].fillna("Unknown")

# Fill numeric columns with 0
for col in df_clean.select_dtypes(include="number").columns:
    df_clean[col] = df_clean[col].fillna(0)

print(df_clean.isna().sum().sum())  # Should be 0
```
