```python
import pandas as pd
```

```python

df1 = pd.DataFrame({'Name':['Anu','Ram']})
df2 = pd.DataFrame({'Name':['Priya','Kiran']})

pd.concat([df1, df2])
```

```python
employees = pd.DataFrame({
    'id':[1,2,3],
    'name':['Anu','Ram','Priya']
})

salaries = pd.DataFrame({
    'id':[2,3,4],
    'salary':[50000,60000,70000]
})

pd.merge(employees, salaries, on='id')
```

```python
pd.merge(employees, salaries, on='id', how='left')
```

```python
pd.merge(employees, salaries, on='id', how='right')
```

```python
pd.merge(employees, salaries, on='id', how='inner')
```

```python
pd.merge(employees, salaries, on='id', how='outer')
```
