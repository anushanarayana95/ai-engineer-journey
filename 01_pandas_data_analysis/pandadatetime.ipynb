```python
df["Open"] = pd.to_numeric(df["Open"], errors="coerce")
df["High"] = pd.to_numeric(df["High"], errors="coerce")
df["Low"] = pd.to_numeric(df["Low"], errors="coerce")
df["Close"] = pd.to_numeric(df["Close"], errors="coerce")
df["Volume"] = pd.to_numeric(df["Volume"], errors="coerce")
```

```python

df['Date'] = pd.to_datetime(
    df['Date'],
    format='%Y-%m-%d %I-%p'
)

df.set_index('Date', inplace=True)

print(type(df.index))
```

```python
df['Close'].resample('W').mean()
```

```python
df['Close'].resample('W').mean().plot(figsize=(12,5))
plt.title("Weekly Average ETH Price")
plt.show()
```

```python
df['Volume'].resample('W').sum().plot(figsize=(12,5))
plt.title("Weekly Trading Volume")
plt.show()
```

```python
df['Close'].hist(bins=30)
plt.title("ETH Price Distribution")
plt.show()
```

```python
df[['Close','7Day_MA','30Day_MA']].plot(figsize=(12,5))
plt.title("Close vs Moving Averages")
plt.show()
```

```python
import matplotlib.pyplot as plt

df['DayOfWeek'] = df.index.day_name()

df.groupby('DayOfWeek')['Close'].mean().plot(kind='bar')

plt.title("Average ETH Close Price by Day")
plt.show()
```
